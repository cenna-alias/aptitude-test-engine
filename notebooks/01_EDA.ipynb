## Step 14: Accuracy Distribution

## Step 15: Response Time Distribution

```python
students["AverageResponseTime"].hist(bins=15)

plt.title("Average Response Time")

plt.xlabel("Seconds")

plt.ylabel("Students")

plt.show()
```

## Step 16: Correlation Matrix

Correlation analysis is performed to understand the relationship between numerical features in the student performance dataset.

```python
import seaborn as sns

corr = students.corr(numeric_only=True)

plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

Observation: The correlation heatmap reveals strong positive relationships between FinalScore, Accuracy, and CorrectAnswers, indicating that better-performing students achieve higher scores. WrongAnswers and AverageResponseTime show strong negative correlations with FinalScore, suggesting that students who make more mistakes or take longer to respond generally score lower. Overall, the heatmap highlights the key factors influencing student performance.
