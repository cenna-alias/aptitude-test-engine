# AptiForge - Model Training

## Objective

The objective of this notebook is to train a Machine Learning model using the preprocessed student performance dataset. A Random Forest Classifier is used to predict the student's aptitude skill level based on their performance features.

## Step 1: Import Required Libraries

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

from sklearn.ensemble import RandomForestClassifier
```

## Step 2: Load Dataset

```python
students = pd.read_csv("../datasets/student_performance.csv")

students.head()
```

## Step 3: Check Dataset Shape

```python
print("Student Dataset Shape")
print(students.shape)
```

## Step 4: Verify Missing Values

```python
students.isnull().sum()
```

## Step 5: Check Duplicate Records

```python
students.duplicated().sum()
```

## Step 6: Select Features and Target Variable

```python
features = [
    "QuestionsAttempted",
    "CorrectAnswers",
    "WrongAnswers",
    "Accuracy",
    "AverageResponseTime",
    "EasyCorrect",
    "MediumCorrect",
    "HardCorrect",
    "LongestCorrectStreak",
    "ConfidenceScore",
    "FinalScore"
]

X = students[features]

y = students["SkillLevel"]

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)
```

## Step 7: Encode Target Variable

```python
encoder = LabelEncoder()

y = encoder.fit_transform(y)

mapping = dict(zip(encoder.classes_, encoder.transform(encoder.classes_)))

print("Class Mapping:")
print(mapping)
```

## Step 8: Split the Dataset into Training and Testing Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

## Step 9: Verify Training and Testing Data

```python
print("Training Features :", X_train.shape)
print("Testing Features :", X_test.shape)

print()

print("Training Labels :", y_train.shape)
print("Testing Labels :", y_test.shape)
```

## Step 10: Create the Random Forest Classifier

Random Forest is an ensemble machine learning algorithm that builds multiple decision trees and combines their predictions to improve classification accuracy and reduce overfitting.

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

## Step 11: Train the Machine Learning Model

```python
model.fit(X_train, y_train)
```

## Step 12: Verify Model Training

The Random Forest model has been successfully trained and is now ready to make predictions on unseen data.

```python
print("Model trained successfully!")
```

## Step 13: Predict Student Skill Levels

```python
y_pred = model.predict(X_test)
```

## Step 14: Compare Actual and Predicted Values

```python
results = pd.DataFrame({
    "Actual Skill Level": y_test,
    "Predicted Skill Level": y_pred
})

print("Total Predictions:", len(results))
results.head(10)
```

## Conclusion

The Random Forest Classifier was successfully trained using the student performance dataset. The model generated predictions for the testing dataset, which will be evaluated in the next notebook using performance metrics such as Accuracy, Precision, Recall, F1-Score, and Confusion Matrix.
